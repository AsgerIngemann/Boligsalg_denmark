```python
import requests
import pandas as pd

res = requests.get(url = "https://api.dataforsyningen.dk/postnumre").json()
```

```python
df = pd.DataFrame(res)
```

```python
df['longitude'] = df['visueltcenter'].apply(lambda x: x[0])
df['latitude'] = df['visueltcenter'].apply(lambda x: x[1])
```

```python
df_out = df[['nr', 'longitude', 'latitude']]
```

```python
df_out.to_csv("postnummer_data.csv", index=False)
```
